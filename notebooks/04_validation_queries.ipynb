#Validation Queries
###1. How many rows arrived with each file
###2. Rows in quarantine
###3. SCD2 recalls with more than one version
###4. Open vs closed recalls 

```sql
USE workspace.food_safety;
```

##Step 1

```sql
SELECT source_file, COUNT(*) AS rows
FROM bronze_recall_events GROUP BY ALL ORDER BY source_file;
```

##Step 2

```sql
SELECT failed_rule, COUNT(*) AS rows
FROM silver_recall_events_quarantine GROUP BY ALL;
```

##Step 3

```sql
SELECT recall_number, status, classification, __START_AT, __END_AT
FROM silver_recalls_history
WHERE recall_number IN (
  SELECT recall_number FROM silver_recalls_history
  GROUP BY recall_number HAVING COUNT(*) > 1 LIMIT 5)
ORDER BY recall_number, __START_AT;
 
```

##Step 4

```sql
SELECT status, COUNT(*) AS recalls 
FROM gold_recalls_current 
GROUP BY ALL;
```
