#Release Next Batch

```python
STAGING = "/Volumes/workspace/food_safety/staging"
LANDING = "/Volumes/workspace/food_safety/landing"
```

```python
dbutils.widgets.text("batches_to_release", "1")
n = int(dbutils.widgets.get("batches_to_release"))
```

```python
staged = sorted(f.name for f in dbutils.fs.ls(STAGING)
                if f.name.startswith("recall_events_"))
released = {f.name for f in dbutils.fs.ls(LANDING)}
pending = [f for f in staged if f not in released]

if not pending:
    dbutils.notebook.exit("No pending batches: All batched released")

for name in pending[:n]:
    dbutils.fs.cp(f"{STAGING}/{name}", f"{LANDING}/{name}")

print("Released:", name)

print(f"Pending batched after this run: {len(pending) - min(n, len(pending))}")
```
