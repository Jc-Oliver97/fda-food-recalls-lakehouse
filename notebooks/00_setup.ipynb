#Schema and Volumes Creation
1. Create the Schema
2. Create 3 Volumes:
    - raw, staging, landing

```sql
CREATE SCHEMA IF NOT EXISTS workspace.food_safety
COMMENT 'FDA Food Recalls Lakehouse';
 
CREATE VOLUME IF NOT EXISTS workspace.food_safety.raw;
 
CREATE VOLUME IF NOT EXISTS workspace.food_safety.staging;
 
CREATE VOLUME IF NOT EXISTS workspace.food_safety.landing;
 
SHOW VOLUMES IN workspace.food_safety;
```
